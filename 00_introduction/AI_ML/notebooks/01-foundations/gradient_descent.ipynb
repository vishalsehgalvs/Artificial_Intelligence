# A price guess that improves

Imagine a shop where two apples should cost $5. Our first guess is $0. That is a mistake of $5. A **weight** is an adjustable price-per-apple number. In the next cell we change that number by one small step, so our guess moves closer to $5.

Before running: starting from a weight of 0, will the next weight be **2** or **-2**? Work it out using the example in [the foundations lesson](../../docs/01-foundations.md). Then run the cell and compare. This notebook runs on your CPU without a model download.

```python
import numpy as np
print('NumPy:', np.__version__)
feature, target, weight, rate = 2.0, 5.0, 0.0, 0.1
gradient = 2 * feature * (weight * feature - target)
weight -= rate * gradient
assert (gradient, weight) == (-20.0, 2.0)
print('gradient:', gradient, 'updated weight:', weight)
```

## Fit a synthetic line
Mean squared error is the average of squared prediction errors. The slope and intercept gradients must both be computed from the **same** old parameters before either update. The loss should fall; that alone does not guarantee generalization.

```python
features = np.linspace(-1, 1, 31)
targets = 2 * features + 1
slope, intercept, losses = 0.0, 0.0, []
for step in range(300):
    errors = slope * features + intercept - targets
    losses.append(float(np.mean(errors ** 2)))
    slope_gradient = 2 * np.mean(features * errors)
    intercept_gradient = 2 * np.mean(errors)
    slope -= 0.1 * slope_gradient
    intercept -= 0.1 * intercept_gradient
assert losses[-1] < losses[0] and abs(slope - 2) < 0.01
assert abs(intercept - 1) < 0.01
print('initial/final loss:', losses[0], losses[-1], 'slope/intercept:', slope, intercept)
```

## Try it
Change the rate to 2. Predict whether the loss will fall. Restore 0.1 afterward. Answer: the updates overshoot and often diverge; a learning rate is not a measure of model quality.
