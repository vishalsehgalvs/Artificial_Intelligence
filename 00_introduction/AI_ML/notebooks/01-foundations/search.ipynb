# Which road gets you home?

Imagine `S` is School and `G` is Home. You can go `S -> A -> G` using two roads that cost 1 and 9, for a total of 10. Or go `S -> B -> C -> G` using three roads costing 2 each, for a total of 6.

**Before running:** Which trip has fewer roads? Which has a lower fare? The first method, breadth-first search (`bfs`), looks for the fewest roads; the second, `cheapest`, adds the fares. Watch the two printed routes disagree. Read [Classical AI](../../docs/02-classical-ai.md) if the queue/heap code is new. This lab needs only Python, no model or downloads.

```python
from collections import deque
from heapq import heappop, heappush
roads = {'S': [('A', 1), ('B', 2)], 'A': [('G', 9)], 'B': [('C', 2)], 'C': [('G', 2)], 'G': []}
def bfs(graph, start, goal):
    queue = deque([(start, [start])])
    seen = {start}
    while queue:
        node, path = queue.popleft()
        if node == goal:
            return path
        for neighbor, cost in graph[node]:
            if neighbor not in seen:
                seen.add(neighbor)
                queue.append((neighbor, path + [neighbor]))
    return None
def cheapest(graph, start, goal):
    frontier = [(0, start, [start])]
    best = {start: 0}
    while frontier:
        total, node, path = heappop(frontier)
        if total > best[node]:
            continue
        if node == goal:
            return total, path
        for neighbor, cost in graph[node]:
            assert cost >= 0
            new_total = total + cost
            if new_total < best.get(neighbor, float('inf')):
                best[neighbor] = new_total
                heappush(frontier, (new_total, neighbor, path + [neighbor]))
    return None
assert bfs(roads, 'S', 'G') == ['S', 'A', 'G']
assert cheapest(roads, 'S', 'G') == (6, ['S', 'B', 'C', 'G'])
assert bfs(roads, 'G', 'missing') is None
print('BFS:', bfs(roads, 'S', 'G'))
print('Cheapest:', cheapest(roads, 'S', 'G'))
```
