# Sort unlabeled points into groups

Imagine a bowl of buttons with no labels: some look similar, so you make piles. **Clustering** makes such groups using measured features, without knowing the right categories. This notebook generates 90 points with four measurements each and asks for three groups.

**Before running:** Do you expect exactly three groups when we explicitly ask k-means for three? **Yes.** That does *not* prove the groups are meaningful. PCA squeezes four measurements into two for easier inspection; the printed variance fraction says how much numeric spread those two directions retain. Read [Machine learning](../../docs/03-machine-learning.md) first. This lab runs locally after dependencies are installed.

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
features, hidden_labels = make_blobs(n_samples=90, centers=3, n_features=4, cluster_std=0.6, random_state=42)
pipeline = make_pipeline(StandardScaler(), PCA(n_components=2, random_state=42))
reduced = pipeline.fit_transform(features)
clusters = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(reduced)
assert reduced.shape == (90, 2) and len(set(clusters)) == 3
print('PCA variance retained:', pipeline[-1].explained_variance_ratio_.sum())
print('silhouette:', silhouette_score(reduced, clusters))
print('first five clusters:', clusters[:5])
```

## Check
Why aren't `hidden_labels` used for training? Clustering is unsupervised. Try three clusters on overlapping blobs and observe that a high silhouette or a pretty 2D plot cannot guarantee real-world semantic categories.
